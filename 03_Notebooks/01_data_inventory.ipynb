```python

import pandas as pd
from pathlib import Path

# Define path to folder
data_path = Path(r"../01_data/raw")

def load_csv (filename):
    return pd.read_csv(data_path/filename)

# Display the files available
for file in sorted(data_path.iterdir()):
    if file.is_file():
        print(file.name)
```

```python

# Load customers table
customers = load_csv("tbl_customers.csv")
customers.head()
```

```python
# Load date table
date = load_csv("tbl_date.csv")
date.head()
```

```python
# Load employees table
employees = load_csv("tbl_employees.csv")
employees.head()
```

```python
# Load products table
products = load_csv("tbl_products.csv")
products.head()
```

```python
# Load stores table
stores = load_csv("tbl_stores.csv")
stores.head()
```

```python
# Load order details table
order_details = load_csv("tbl_order_details.csv")
order_details.head()
```

```python
# Load 2022-2023 orders table
orders_2022_2023 = load_csv("tbl_orders_2022_2023.csv")
orders_2022_2023.head()
```

```python
# Load 2024 orders table
orders_2024 = load_csv("tbl_orders_2024.csv")
orders_2024.head()
```

```python
# Load returns table
returns = load_csv("tbl_returns.csv")
returns.head()
```

### Data Inventory

With all tables loaded, we can now create a data inventory that documents the actual structure of each table

```python

datasets = {
    "customers": customers,
    "date": date,
    "employees": employees,
    "products": products,
    "stores": stores,
    "order_details": order_details,
    "orders_2022_2023": orders_2022_2023,
    "orders_2024": orders_2024,
    "returns": returns
}

for name, df in datasets.items():
    print(f"\n{'=' * 50}")
    print(f"Table: {name}")
    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")
    print("Column names:")
    print(df.columns.tolist())
```
